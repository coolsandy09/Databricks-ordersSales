#### READ `GEOLOCATIONS` FROM VOLUME

```python
BASE_DIR = '/Volumes/sales/landing/sales/GeoLocations/'
CATALOG = 'SALES'
SCHEMA = 'BRONZE'
TARGET_TABLE= 'GeoLocations'
```

```python
geo_schema = '''geolocation_zip_code_prefix INT,
               geolocation_lat DOUBLE,
               geolocation_lng DOUBLE,
               geolocation_city string,
               geolocation_state string'''
```

```python
geo_df=spark.read.format('csv')\
                    .option('header',True)\
                    .schema(geo_schema)\
                    .load(f'{BASE_DIR}')
display(geo_df.limit(3))
geo_df.printSchema()
```

```python
from pyspark.sql.functions import *
geo_trans_df = geo_df.select('*',"_metadata.file_name","_metadata.file_path",
                               "_metadata.file_modification_time",current_timestamp().alias('ingestion_timestamp'))

display(geo_trans_df.limit(1))
```

```python
geo_trans_df.write.mode('append').format('delta').saveAsTable(f'{CATALOG}.{SCHEMA}.{TARGET_TABLE}')
```
