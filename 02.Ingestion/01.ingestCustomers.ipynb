#### Read Customers From Volume

```python
Base_Dir = '/Volumes/sales/landing/sales/Customers/'
Catalog = 'Sales'
Schema = 'bronze'
target_table = 'customers'
```

```python
customers_schema = """customer_id STRING, 
                 customer_unique_id STRING, 
                 customer_zip_code_prefix INT, 
                 customer_city STRING, 
                 customer_state STRING
                 """
```

```python
cust_schema ='''StructType([ 
  StructField('customer_id', StringType(), True),
  StructField('customer_unique_id', StringType(), True),
  StructField('customer_zip_code_prefix', IntegerType(), True),
  StructField('customer_city', StringType(), True),
  StructField('customer_state', StringType(), True)
])'''
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
cust_df=spark.read.format('csv').\
                    option('header',True)\
                    .schema(eval(cust_schema))\
                    .load(f'{Base_Dir}')
display(cust_df.limit(3))
cust_df.printSchema()
```

```python
cust_df.write.mode('append').format('delta').saveAsTable(f'{Catalog}.{Schema}.{target_table}')
```

```python
cust_trans_df=cust_df.select('*',"_metadata.file_name","_metadata.file_path","_metadata.file_modification_time")
display(cust_trans_df)
```
