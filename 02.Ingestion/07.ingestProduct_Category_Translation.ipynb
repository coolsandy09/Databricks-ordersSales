#### READ `Product_Category_Translation` FROM VOLUME

```python
BASE_DIR = '/Volumes/sales/landing/sales/Product_Category_Translation/'
CATALOG = 'SALES'
SCHEMA = 'BRONZE'
TARGET_TABLE = 'Product_Category_Translation'
```

```python
PCT_schema = '''product_category_name string,
                product_category_name_english string'''
```

```python
PCT_df = spark.read.format('csv')\
                   .option('header', True)\
                   .schema(PCT_schema)\
                   .load(f'{BASE_DIR}')
display(PCT_df.limit(1))
PCT_df.printSchema()
```

```python
from pyspark.sql.functions import *
geo_trans_df = PCT_df.select('*',"_metadata.file_name","_metadata.file_path",
                               "_metadata.file_modification_time",current_timestamp().alias('ingestion_timestamp'))

display(geo_trans_df.limit(1))
```

```python
geo_trans_df.write.mode('append').format('delta').saveAsTable(f'{CATALOG}.{SCHEMA}.{TARGET_TABLE}')
```
