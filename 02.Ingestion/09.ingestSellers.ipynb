####READ `Sellers` FROM VOLUMES


```python
BASE_DIR = '/Volumes/sales/landing/sales/Sellers/'
CATALOG = 'SALES'
SCHEMA = 'BRONZE'
TARGET_TABLE= 'Sellers'
```

```python
Sellers_schema = '''seller_id STRING, 
                 seller_zip_code_prefix INT, 
                 seller_city STRING, 
                 seller_state STRING'''
```

```python
Sellers_df=spark.read.format('csv')\
                    .option('header',True)\
                    .schema(Sellers_schema)\
                    .load(f'{BASE_DIR}')
display(Sellers_df.limit(3))
Sellers_df.printSchema()
```

```python
from pyspark.sql.functions import *
Sellers_trans_df = Sellers_df.select('*',"_metadata.file_name","_metadata.file_path",
                               "_metadata.file_modification_time",current_timestamp().alias('ingestion_timestamp'))

display(Sellers_trans_df.limit(1))
```

```python
Sellers_trans_df.write.mode('append').format('delta').saveAsTable(f'{CATALOG}.{SCHEMA}.{TARGET_TABLE}')
```
